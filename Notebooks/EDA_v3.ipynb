### Conclusiones — EDA de población INEGI

Se analizaron los datos del Censo de Población y Vivienda 2020 correspondientes a Ciudad de México.

Se identificaron 66,456 registros de manzanas urbanas, con identificadores geográficos únicos y sin valores no numéricos en la variable de población total (`POBTOT`).

La población agregada de las manzanas es de 9,145,590 habitantes, equivalente al 99.30% de la población censal total de CDMX.

Se identificaron 3,373 manzanas sin habitantes (5.08%), las cuales se conservarán en el análisis.

**Decisión metodológica:** utilizar la población a nivel de manzana para estimar la cobertura residencial actual y adicional de ECOBICI, reconociendo la población no representada y las limitaciones temporales del Censo 2020.

```python
import geopandas as gpd

geo_manzanas = gpd.read_file(
    "../Data/INEGI/poligonos.json",
    rows=5
)

geo_manzanas.head()
```

```python
manzanas["CVEGEO"] = (
    manzanas["ENTIDAD"] +
    manzanas["MUN"] +
    manzanas["LOC"] +
    manzanas["AGEB"] +
    manzanas["MZA"]
)

manzanas[["CVEGEO", "POBTOT_num"]].head()
```

```python
geo_manzanas = gpd.read_file("../Data/INEGI/poligonos.json")

print("Polígonos:", len(geo_manzanas))
print("IDs duplicados:", geo_manzanas["CVEGEO"].duplicated().sum())
```

```python
ids_censo = set(manzanas["CVEGEO"])
ids_geo = set(geo_manzanas["CVEGEO"])

print("Manzanas censales:", len(ids_censo))
print("Polígonos geográficos:", len(ids_geo))

print("Coincidencias:", len(ids_censo & ids_geo))
print("Sin geometría:", len(ids_censo - ids_geo))
print("Polígonos sin datos censales:", len(ids_geo - ids_censo))
```

```python
sin_geometria = manzanas[
    ~manzanas["CVEGEO"].isin(geo_manzanas["CVEGEO"])
]

print("Población sin geometría:", sin_geometria["POBTOT_num"].sum())

sin_geometria[["CVEGEO", "NOM_MUN", "POBTOT_num"]]
```

```python
poblacion_geo = geo_manzanas.merge(
    manzanas[["CVEGEO", "POBTOT_num"]],
    on="CVEGEO",
    how="inner",
    validate="one_to_one"
)

poblacion_geo.head()
```

```python
print("Manzanas:", len(poblacion_geo))
print("Población total:", poblacion_geo["POBTOT_num"].sum())
print("CRS:", poblacion_geo.crs)

poblacion_geo.plot(
    column="POBTOT_num",
    figsize=(10, 10),
    legend=True
)
```

```python
poblacion_geo = poblacion_geo.to_crs(epsg=32614)

# Representar cada manzana mediante un punto interior
puntos_poblacion = poblacion_geo.copy()
puntos_poblacion["geometry"] = poblacion_geo.representative_point()
```

```python
estaciones = pd.read_csv(
    "../Data/Estaciones_ECOBICI/estaciones_ECOBICI.csv",
    encoding="latin-1",
    dtype={"num_cicloe": "string"}
).dropna(how="all")

estaciones_geo = gpd.GeoDataFrame(
    estaciones,
    geometry=gpd.points_from_xy(
        estaciones["longitud"],
        estaciones["latitud"]
    ),
    crs="EPSG:4326"
).to_crs(epsg=32614)

estaciones_geo.head()
```

```python
buffers = estaciones_geo[["num_cicloe", "geometry"]].copy()

buffers["geometry"] = buffers.buffer(500)

buffers.plot(figsize=(10, 8), alpha=0.3)
```

```python
# Identificar población dentro de los buffers de ECOBICI
poblacion_cubierta = gpd.sjoin(
    puntos_poblacion,
    buffers,
    how="inner",
    predicate="within"
)

poblacion_cubierta.head()
```

```python
# Eliminar manzanas repetidas
poblacion_cubierta = poblacion_cubierta.drop_duplicates(
    subset="CVEGEO"
)

# Calcular cobertura
habitantes_cubiertos = poblacion_cubierta["POBTOT_num"].sum()
habitantes_totales = puntos_poblacion["POBTOT_num"].sum()

porcentaje_cobertura = (
    habitantes_cubiertos / habitantes_totales * 100
)

print(f"Habitantes cubiertos: {habitantes_cubiertos:,.0f}")
print(f"Habitantes sin cobertura: {habitantes_totales - habitantes_cubiertos:,.0f}")
print(f"Cobertura poblacional: {porcentaje_cobertura:.2f}%")
```

```python
puntos_poblacion["cubierta"] = (
    puntos_poblacion["CVEGEO"].isin(poblacion_cubierta["CVEGEO"])
)

fig, ax = plt.subplots(figsize=(10, 9))

puntos_poblacion.plot(
    ax=ax,
    column="cubierta",
    categorical=True,
    markersize=1,
    legend=True
)

plt.title("Cobertura poblacional actual de ECOBICI (500 m)")
plt.show()
```

```python
puntos_poblacion[
    ["CVEGEO", "POBTOT_num", "cubierta"]
].to_csv(
    "../Data/Procesados/cobertura_poblacional_ECOBICI.csv",
    index=False
)
```

### Conclusiones — Cobertura poblacional actual de ECOBICI

Se estimó la cobertura poblacional residencial del sistema ECOBICI mediante radios de 500 metros alrededor de las 677 estaciones del catálogo geográfico.

El análisis utilizó 66,449 manzanas georreferenciadas del Censo 2020, que representan una población total de 9,145,155 habitantes.

**Resultados:**
- Población cubierta: 1,167,905 habitantes.
- Población sin cobertura: 7,977,250 habitantes.
- Cobertura poblacional estimada: 12.77%.

Para evitar el doble conteo, cada manzana se contabilizó una sola vez, independientemente del número de estaciones cercanas.

**Limitaciones:** la estimación utiliza distancias euclidianas, población censal de 2020 y un punto representativo por manzana. No incorpora distancias reales de caminata ni desplazamientos cotidianos de la población.

Este resultado constituye la cobertura inicial contra la cual se evaluará el incremento poblacional generado por nuevas estaciones.
